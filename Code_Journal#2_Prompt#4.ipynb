```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#1000 random numbers using Cauchy Distribution
x = np.random.standard_cauchy(size=1000)

#plotting the bins
plt.hist(x, bins=100, alpha = 0.5, edgecolor = "black")
plt.ylabel("N per bin")
plt.xlabel("x")
plt.savefig("Code_Journal#2_Prompt#4.pdf", bbox_inches = "tight", facecolor = "white")
```
