```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def x_array():
    x = np.linspace(0, 1, 100) #x array ranging from 0-1 in 100 steps

    return x, np.exp(x)
```

```python
x, y = x_array()

fig = plt.figure(figsize=(5,5)) #create a 5x5 image

#subplot and plot x vs y
ax = fig.add_subplot(1,1,1)
ax.plot(x, y)

#label the axes
ax.set_xlabel('Time [milliseconds]') #add an x label
ax.set_ylabel('Awesomeness') #add a y label

#save the plot
plt.savefig("contour-surface.pdf",bbox_inches="tight",facecolor="white")
```
