```python
#Calculates sin(x)
```

```python
import numpy as np

def calc_sin(x):
    return np.sin(x)
```

```python
#Calculates cos(x)
```

```python
def calc_cos(x):
    return np.cos(x)
```

```python
#Tabulates between x values, sin(x) values, and cos(x) values
```

```python
x_values = np.linspace(0, 2 * np.pi, 1000)

sin_values = calc_sin(x_values)
cos_values = calc_cos(x_values)
```

```python
#Prints these values in 10 respective columns
```

```python
print(f"{'x':<15} {'sin(x)':<15} {'cos(x)':<15}")

for n in range(10):
    print(f"{x_values[n]:<15.6f} {sin_values[n]:<15.6f} {cos_values[n]:<15.6f}")
```
