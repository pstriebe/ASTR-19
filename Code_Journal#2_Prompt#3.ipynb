```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#1000 random numbers distributed uniformly between [0,1]
x = np.random.rand(1000)

#plotting the bins
plt.hist(x, bins=100, alpha = 0.5, edgecolor = "black")
plt.ylabel("N per bin")
plt.xlabel("x")
plt.savefig("Code_Journal#2_Prompt#3.png", bbox_inches = "tight", facecolor = "white")
```
